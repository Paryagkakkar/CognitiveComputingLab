```python
#4 1st part
r = 1024170447

A = set(str(r))
B = {0, 2, 4, 6, 8}

A = {int(i) for i in A}

print("A =", A)
print("B =", B)
#2nd part
print("Union =", A | B)
# 3rd
print("Intersection =", A & B)
# 4th
print("A - B =", A - B)
print("B - A =", B - A)
#5th
print("Symmetric difference =", A ^ B)
#6th
print("A subset of B =", A.issubset(B))
print("B superset of A =", B.issuperset(A))
```

```python
#5 1st

d = {
    "name": "Paryag",
    "roll": 1024170447,
    "branch": "COPC",
    "age": 19,
    "city": "Sirsa"
}

a = d.pop("city")
d["location"] = a

print("After changing city to location =", d)


# 2nd

d["cgpa"] = 7

print("After adding cgpa =", d)


# 3rd

d["age"] = d["age"] + 1

print("After increasing age =", d)


# 4th

d2 = d.copy()

a = d2.pop("branch")

print("After pop =", d2)
print("Removed value =", a)


d3 = d.copy()

del d3["branch"]

print("After del =", d3)

# pop() removes the key and returns its value, while del only removes the key and does not return its value.


# 5th

for i, j in d.items():
    print(i, "→", j)


# 6th
if "email" in d:
    print("Email =", d["email"])
else:
    print("Email key does not exist")


# 7th

friend_dict = {
    "name": "Rahul",
    "roll": 1234567890,
    "branch": "CSE",
    "age": 20,
    "city": "Delhi"
}

d4 = {**d, **friend_dict}

print("Merged dictionary =", d4)

# If both dictionaries have the same key, the value from the second dictionary wins.


#8th

d5 = {i: j for i, j in d.items() if isinstance(j, str)}

print("String values dictionary =", d5)
```
